Посмотрели полную таблицу (10 рандомных примеров), а также проверим датасет на полные дубли

```python
print("Полные дубликаты:", df.duplicated().sum())
print("Дубликаты по student_id:", df.duplicated(subset=['student_id']).sum())
```

Получили 37 дублей, удалим их

```python
df = df.drop_duplicates()
print("Полные дубликаты:", df.duplicated().sum())
```

Здесь видно первые грубы ошибки - некоторые города находятся под другими наименованиями. С помощью словаря распределим эти города по нужным названиями 

```python
for col in ['region', 'device_type', 'grade']:
    print(df[col].value_counts(dropna=False))

print(df['grade'].value_counts().sort_index())
```

```python
region_map = {
    # Казань
    'г.Казань': 'Казань',
    'казань':   'Казань',
    # Санкт-Петербург
    'СПб':            'Санкт-Петербург',
    'спб':            'Санкт-Петербург',
    'Санкт Петербург':'Санкт-Петербург',
    # Москва
    'мск':      'Москва',
    'москва':   'Москва',
    'г. Москва':'Москва',
    'Moskva':   'Москва',
}

df['region'] = df['region'].replace(region_map)

print(df['region'].value_counts())
assert df['region'].nunique() == 10
assert set(df['region'].unique()) == {
    'Москва','Санкт-Петербург','Казань','Екатеринбург','Новосибирск',
    'Ростов-на-Дону','Краснодар','Воронеж','Пермь','Уфа'
}
```

После проведения маппинга, проверим еще раз device_type и grade

```python
print(df['device_type'].value_counts(dropna=False))
print(df['grade'].value_counts().sort_index())
```

Теперь получилось ровно 8000 уникальных учеников

```python
print(df.shape) 
```

Посчитаем длительность 

```python
df['end_time'].dtype
df['start_time'] = pd.to_datetime(df['start_time'], errors='coerce')
df['end_time']   = pd.to_datetime(df['end_time'], errors='coerce')
print("Не распарсилось start_time:", df['start_time'].isna().sum())
print("Не распарсилось end_time:  ", df['end_time'].isna().sum())

df['duration_min'] = (df['end_time'] - df['start_time']).dt.total_seconds() / 60

print(df['duration_min'].describe())
print("Отрицательные длительности:", (df['duration_min'] < 0).sum())
print("NaN длительности:", df['duration_min'].isna().sum())
```

```python
print(df['school_number'].value_counts().head(15))
print("Всего уникальных школ:", df['school_number'].nunique())
```

```python
fast = df[df['duration_min'] < 3].sort_values('duration_min')
print(f"Сессий < 3 мин: {len(fast)}")
print(fast[['student_id', 'region', 'grade', 'total_score', 
            'duration_min', 'device_type', 'attempts_count', 'disconnects']])

slow = df[df['duration_min'] > 19].sort_values('duration_min')
print(f"Сессий > 19 мин: {len(slow)}")
print(slow[['student_id', 'region', 'grade', 'total_score', 
            'duration_min', 'device_type', 'attempts_count', 'disconnects']])
```
